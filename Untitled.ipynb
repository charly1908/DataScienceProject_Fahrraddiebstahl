```python
import pandas as pd
import matplotlib . pyplot as plt
import seaborn as sns

df = pd.read_csv('Fahrraddiebstahl.csv', sep=',', encoding='latin1')
print(df.info())
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print("Schaden:", df['SCHADENSHOEHE'].mean())
```

```python
#Was ist der durchschnittliche Schaden für die verschiedenen Fahrradarten?
df_GroupArtMean = df.groupby('ART_DES_FAHRRADS')['SCHADENSHOEHE'].mean()
df_GroupArtMean = df_GroupArtMean.sort_values(ascending=True)

fig , ax = plt . subplots ()
ax.spines[ 'top' ].set_visible( False )
ax.spines[ 'right' ].set_visible( False )
ax = sns.barplot(y=df_GroupArtMean.index, x=df_GroupArtMean.values, color='blue', edgecolor='black')

ax.set_xlim(0, 4000)
ax.set_xticks(range(0, 4001, 500))
ax.set_xlabel( xlabel='Schaden' , fontsize=12)
ax.set_ylabel( ylabel='Fahrradtyp' , fontsize=12)
ax.set_title( label='Schaden pro Radtyp (Avg)' ,fontsize=20)
fig.set_size_inches(12, 8, forward=True)
plt.savefig('SchadenRadtyp.pdf', bbox_inches='tight')
plt.show()
```

```python
#Was ist die durchschnittliche Standardabweichung für die verschiedenen Fahrradarten?

df_GroupArtStd = df.groupby('ART_DES_FAHRRADS')['SCHADENSHOEHE'].std()
df_GroupArtStd = df_GroupArtStd.sort_values(ascending=True)

fig , ax = plt . subplots ()
ax.spines[ 'top' ].set_visible( False )
ax.spines[ 'right' ].set_visible( False )
ax = sns.barplot(y=df_GroupArtStd.index, x=df_GroupArtStd.values, color='blue', edgecolor='black')

ax.set_xlim(0, 4000)
ax.set_xticks(range(0, 4001, 500))
ax.set_xlabel( xlabel='Schaden' , fontsize=12)
ax.set_ylabel( ylabel='Fahrradtyp' , fontsize=12)
ax.set_title( label='Abweichung Schaden pro Radtyp' ,fontsize=20)
fig.set_size_inches(12, 8, forward=True)
plt.savefig('SchadenRadtypStd.pdf', bbox_inches='tight')
plt.show()
```

```python
df_GroupArtStability = (df_GroupArtStd / df_GroupArtMean) * 100

df_GroupArtStability = df_GroupArtStability.sort_values(ascending=True)

fig, ax = plt.subplots()

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

sns.barplot(
    y=df_GroupArtStability.index,
    x=df_GroupArtStability.values,
    color='blue',
    edgecolor='black',
    ax=ax
)

ax.set_xlim(0, 130)
ax.set_xticks(range(0, 131, 10))

ax.set_xlabel('Standardabweichung in %', fontsize=12)
ax.set_ylabel('Fahrradtyp', fontsize=12)
ax.set_title('Relative Standardabweichung pro Radtyp', fontsize=20)

fig.set_size_inches(12, 8)

plt.savefig('SchadenRadtypStd.pdf', bbox_inches='tight')
plt.show()
```

```python
# Was wird am häufigsten geklaut?
df_GroupArtCount = df['ART_DES_FAHRRADS'].value_counts().sort_values(ascending=True)

fig, ax = plt.subplots()

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

sns.barplot(
    y=df_GroupArtCount.index,
    x=df_GroupArtCount.values,
    color='blue',
    edgecolor='black',
    ax=ax
)

ax.set_xlabel('Anzahl der Diebstähle', fontsize=12)
ax.set_ylabel('Fahrradtyp', fontsize=12)
ax.set_title('Fahrraddiebstähle nach Fahrradtyp', fontsize=20)

fig.set_size_inches(12, 8)

plt.savefig('SchadenRadtyp.pdf', bbox_inches='tight')
plt.show()
```
