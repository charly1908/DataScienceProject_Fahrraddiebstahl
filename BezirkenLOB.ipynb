```python
#Wir möchten die beiden Datensätze über eine gemeinsame Kennung verbinden:

#Fahrraddiebstahl (LOR) → Bezirksübersicht (PLR_ID) → Bezirk
# IDs als Text speichern
diebstahl["LOR"] = diebstahl["LOR"].astype("string").str.zfill(8)
bezirke["PLR_ID"] = bezirke["PLR_ID"].astype("string").str.zfill(8)
#LOR	Gibt an, in welchem kleinen Gebiet (Planungsraum) das Fahrrad gestohlen wurde.
#PLR_ID	Ist die ID dieses kleinen Gebiets in deiner Bezirksübersicht.
#BEZ_ID	Ist die ID des gesamten Berliner Bezirks, z. B. Mitte.
# Tabellen verbinden
daten_join = pd.merge(
    diebstahl, #tabelle links
    bezirke[["PLR_ID", "BEZ"]],#tabelle rechts
    left_on="LOR",# LOR wird verwendet
    right_on="PLR_ID",
    how="left",
    validate="many_to_one"
)

daten_join.head()
#| Code  Bedeutung 
# `pd.merge()`  Verbindet zwei Tabellen 
#`left_on="LOR"`  Verbindungsspalte der Diebstahltabelle 
# `right_on="PLR_ID"` Verbindungsspalte der Bezirksabelle 
#`how="left"`  Alle Diebstähle bleiben erhalten 
# `validate="many_to_one"`  Mehrere Diebstähle dürfen zu einem Planungsraum gehören 
#Verwende die Spalte LOR aus der linken Tabelle, um die Verbindung herzustellen
```

```python
print(daten_join[["LOR", "BEZ"]].head())

print("Nicht zugeordnete Diebstähle:")
print(daten_join["BEZ"].isna().sum())
#Jeder zuordenbare Fahrraddiebstahl bekommt einen Bezirksnamen.
```

```python
#Zuordnung, um herauszufinden, in welchem Bezirk die einzelnen Fahrraddiebstähle stattgefunden haben.
```

```python
diebstahl_pro_bezirk = (
    daten_join.groupby("BEZ")
    .size()
    .reset_index(name="Anzahl_Diebstaehle")
)

diebstahl_pro_bezirk
#groupby("BEZ")	Gruppiert alle Diebstähle nach Bezirk
#reset_index()	Erstellt daraus eine normale Tabelle
```

```python
daten_join["Datum"] = pd.to_datetime(
    daten_join["TATZEIT_ANFANG_DATUM"],
    dayfirst=True,
    errors="coerce"
)

daten_join["Monat"] = daten_join["Datum"].dt.month
#
```

```python
monatliche_diebstaehle = pd.crosstab(
    daten_join["BEZ"],
    daten_join["Monat"]
)

monatliche_diebstaehle
#Diebstähle pro Bezirk und Monat zählen
#Ziel „Welche Berliner Bezirke weisen ähnliche saisonale Muster bei Fahrraddiebstählen auf?
```

```python
# Sicherstellen, dass alle 12 Monate vorhanden sind
monatliche_diebstaehle = monatliche_diebstaehle.reindex(
    columns=range(1, 13),
    fill_value=0
)

# Monatswerte in Prozent umrechnen
X = monatliche_diebstaehle.div(
    monatliche_diebstaehle.sum(axis=1),
    axis=0
) * 100

X
#in welchen Monaten die Fahrraddiebstähle in den Bezirken ähnlich verteilt sind.
#Cluster 0: Mitte und Neukölln
#- Ähnliche monatliche Verteilung.
#- Die Diebstahlanteile steigen von Januar bis April.
#Cluster 1: Spandau und Reinickendorf
##- Ebenfalls ähnliche monatliche Verteilung.
#- Die Diebstahlanteile sinken von Januar bis April.
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

cluster = kmeans.fit_predict(X)
```

```python
ergebnis = X.copy()

ergebnis["Cluster"] = cluster

ergebnis
```

```python
ergebnis[["Cluster"]].sort_values("Cluster")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

# Jeden Bezirk als eigene Linie darstellen
for bezirk in X.index:
    cluster = ergebnis.loc[bezirk, "Cluster"]

    plt.plot(
        X.columns,
        X.loc[bezirk],
        marker="o",
        label=f"{bezirk} (Cluster {cluster})"
    )

plt.xlabel("Monat")
plt.ylabel("Anteil der Diebstähle (%)")
plt.title("Monatliche Fahrraddiebstahlmuster der Berliner Bezirke")

plt.xticks(range(1, 13))
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Durchschnittliche Monatswerte pro Cluster
cluster_mittelwerte = ergebnis.groupby("Cluster")[list(range(1, 13))].mean()

plt.figure(figsize=(12, 6))

for cluster in cluster_mittelwerte.index:
    plt.plot(
        cluster_mittelwerte.columns,
        cluster_mittelwerte.loc[cluster],
        marker="o",
        label=f"Cluster {cluster}"
    )

plt.xlabel("Monat")
plt.ylabel("Durchschnittlicher Diebstahlanteil (%)")
plt.title("Vergleich der monatlichen Diebstahlmuster nach Cluster")

plt.xticks(range(1, 13))
plt.legend()
plt.grid(True)

plt.show()
```
